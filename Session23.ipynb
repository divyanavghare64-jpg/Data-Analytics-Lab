```python
import pandas as pd
import numpy as np

from sklearn.datasets import fetch_california_housing, load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    r2_score, accuracy_score, precision_score,
    recall_score, f1_score, confusion_matrix, classification_report
)
```

```python
 # Q1. (Regression Dataset)

print("--- Question 1: Linear Regression ---")

# Loading California Housing dataset
housing = fetch_california_housing(as_frame=True)
df_housing = housing.frame

# Preprocessing: checking missing values
print("Total missing values in dataset:", df_housing.isnull().sum().sum())

X1 = housing.data
y1 = housing.target

# Splitting train and test data
X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y1, test_size=0.2, random_state=42
)

# Feature Scaling
scaler1 = StandardScaler()
X1_train_scaled = scaler1.fit_transform(X1_train)
X1_test_scaled = scaler1.transform(X1_test)

# Model Training
lr_model = LinearRegression()
lr_model.fit(X1_train_scaled, y1_train)

# Evaluation
y1_pred = lr_model.predict(X1_test_scaled)
print("R2 Score:", round(r2_score(y1_test, y1_pred), 4))
print("Sample Predictions (First 5):", np.round(y1_pred[:5], 2))
```

```python
# Preprocessing for Classification (Q2, Q3, Q4, Q5)

cancer = load_breast_cancer()
X2 = cancer.data
y2 = cancer.target

# Train-test split for classification dataset
X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, random_state=42
)

# Scaling features
scaler2 = StandardScaler()
X2_train_scaled = scaler2.fit_transform(X2_train)
X2_test_scaled = scaler2.transform(X2_test)
```

```python
# Q2. Logistic Regression

print("\n--- Question 2: Logistic Regression ---")

log_reg = LogisticRegression()
log_reg.fit(X2_train_scaled, y2_train)
y2_pred_log = log_reg.predict(X2_test_scaled)

print("Confusion Matrix:\n", confusion_matrix(y2_test, y2_pred_log))
print("Accuracy:", round(accuracy_score(y2_test, y2_pred_log), 4))
print("Precision:", round(precision_score(y2_test, y2_pred_log), 4))
print("Recall:", round(recall_score(y2_test, y2_pred_log), 4))
print("F1 Score:", round(f1_score(y2_test, y2_pred_log), 4))
```

```python
# Q3. K-Nearest Neighbors (KNN)

print("\n--- Question 3: KNN ---")

k_list = [3, 5, 7]
best_k = 3
best_acc = 0

# Checking different k values
for k in k_list:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X2_train_scaled, y2_train)
    acc = knn.score(X2_test_scaled, y2_test)
    print(f"Accuracy for k = {k}: {round(acc, 4)}")

    if acc > best_acc:
        best_acc = acc
        best_k = k

print(f"Best k selected: {best_k}")

# Final prediction with best k
knn_best = KNeighborsClassifier(n_neighbors=best_k)
knn_best.fit(X2_train_scaled, y2_train)
y2_pred_knn = knn_best.predict(X2_test_scaled)
```

```python
# Q4. Naive Bayes

print("\n--- Question 4: Naive Bayes ---")

nb_model = GaussianNB()
nb_model.fit(X2_train_scaled, y2_train)
y2_pred_nb = nb_model.predict(X2_test_scaled)

print("Confusion Matrix:\n", confusion_matrix(y2_test, y2_pred_nb))
print("\nClassification Report:\n", classification_report(y2_test, y2_pred_nb))
```

```python
# Q5. Algorithm Comparison

print("\n--- Question 5: Comparison Table ---")

table_data = {
    'Algorithm': ['Logistic Regression', f'KNN (k={best_k})', 'Naive Bayes'],
    'Accuracy': [
        accuracy_score(y2_test, y2_pred_log),
        accuracy_score(y2_test, y2_pred_knn),
        accuracy_score(y2_test, y2_pred_nb)
    ],
    'Precision': [
        precision_score(y2_test, y2_pred_log),
        precision_score(y2_test, y2_pred_knn),
        precision_score(y2_test, y2_pred_nb)
    ],
    'Recall': [
        recall_score(y2_test, y2_pred_log),
        recall_score(y2_test, y2_pred_knn),
        recall_score(y2_test, y2_pred_nb)
    ],
    'F1 Score': [
        f1_score(y2_test, y2_pred_log),
        f1_score(y2_test, y2_pred_knn),
        f1_score(y2_test, y2_pred_nb)
    ]
}

comparison_df = pd.DataFrame(table_data)
print(comparison_df.to_string(index=False))
```
