```python
# Q1
import pandas as pd
from sklearn.model_selection import train_test_split

# Data loading
df = pd.read_csv('train.txt', sep=';', header=None, names=['text', 'emotion'])

# Separating features (X) and target (y)
X = df['text']
y = df['emotion']

# Perform train-test split with test_size=0.20 and random_state=42. (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Print the shape of X_train, X_test, y_train, and y_test.
print("--- Shapes of Split Data ---")
print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```

```python
# Q2
from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer
cv_bow = CountVectorizer()

# Fit and transform on training data, transform test data
X_train_bow = cv_bow.fit_transform(X_train)
X_test_bow = cv_bow.transform(X_test)

# Print the shape of the resulting matrices
print("X_train_bow shape:", X_train_bow.shape)
print("X_test_bow shape :", X_test_bow.shape)

#  Display the first 20 feature names (Vocabulary words)
vocab = cv_bow.get_feature_names_out()
print("\nFirst 20 feature names:")
print(vocab[:20])
```

```python
# Q3
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Initialize & fit model
model_bow = MultinomialNB()
model_bow.fit(X_train_bow, y_train)

# make Predictions on the test set.
y_pred_bow = model_bow.predict(X_test_bow)

# Calculate and print the accuracy score.
acc_bow = accuracy_score(y_test, y_pred_bow)
print(f"Accuracy with Basic BoW: {acc_bow * 100:.2f}%")
```

```python
# Q4
# Total size of vocabulary
print("Total Vocabulary Size:", len(vocab))

# show any 15 words from vocabulary
print("\nRandom 15 words from vocabulary:")
print(vocab[150:165])

# Convert one sample document into BoW vector
sample_text = [X_train.iloc[0]]
sample_vector = cv_bow.transform(sample_text)

print("\nSample Original Text:")
print(sample_text[0])
print("\nBoW Vector Array Representation (non-zero elements):")
print(sample_vector.toarray())
```

```python
# Q5
# CountVectorizer with Bigrams (unigrams + bigrams)
cv_bigram = CountVectorizer(ngram_range=(1, 2))

X_train_bigram = cv_bigram.fit_transform(X_train)
X_test_bigram = cv_bigram.transform(X_test)

print("Bigram Matrix shape:", X_train_bigram.shape)

# Display some of the bigram features.
bigram_features = cv_bigram.get_feature_names_out()
print("\nSome Bigram Features:")
print(bigram_features[500:515])
```

```python
# Q6
# Train MultinomialNB model using the bigram features created in Q5
model_bigram = MultinomialNB()
model_bigram.fit(X_train_bigram, y_train)

# Making predictions on test data
y_pred_bigram = model_bigram.predict(X_test_bigram)

# Calculate accuracy
acc_bigram = accuracy_score(y_test, y_pred_bigram)

print(f"Accuracy with Bigrams (Unigram + Bigram): {acc_bigram * 100:.2f}%")

# Compare with basic BoW accuracy from Q3
print("\n--- Accuracy Comparison ---")
print(f"Basic BoW Accuracy   : {acc_bow * 100:.2f}%")
print(f"Bigram BoW Accuracy  : {acc_bigram * 100:.2f}%")
```

```python
# Q7
from sklearn.feature_extraction.text import TfidfVectorizer

# Applying TfidfVectorizer on the same train-test split
tfidf = TfidfVectorizer()

# Fit and transform X_train.
X_train_tfidf = tfidf.fit_transform(X_train)

# Transform X_test.
X_test_tfidf = tfidf.transform(X_test)

# Print the shape of the TF_IDF matrices.
print("Shape of X_train_tfidf:", X_train_tfidf.shape)
print("Shape of X_test_tfidf :", X_test_tfidf.shape)

# Display the first 15 feature names
tfidf_words = tfidf.get_feature_names_out()
print("\nFirst 15 TF-IDF Feature Names:")
print(tfidf_words[:15])
```

```python
# Q8
# Train Multinomial Naive Bayes using TF-IDF features
model_tfidf = MultinomialNB()
model_tfidf.fit(X_train_tfidf, y_train)

# Make predictions on the test set.
y_pred_tfidf = model_tfidf.predict(X_test_tfidf)

# Calculate and print accuracy score
acc_tfidf = accuracy_score(y_test, y_pred_tfidf)
print(f"Accuracy with TF-IDF Features: {acc_tfidf * 100:.2f}%")
```

```python
# Q9
import pandas as pd

# Creating a comparison table of the 3 approaches:
results = {
    'Vectorization Method': ['Bag of Words (Unigram)', 'Bag of Words (Unigram + Bigram)', 'TF-IDF'],
    'Accuracy Score': [acc_bow, acc_bigram, acc_tfidf]
}

df_results = pd.DataFrame(results)
print("Final Comparison Table")
print(df_results)
```

Observation:
TF-IDF method works best for text classification because it reduces the importance of very common words (like 'is', 'the') and highlights unique words that carry strong emotion context. Adding bigrams increases feature size greatly, which sometimes introduces noise depending on the text length.

```python
# Q10
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# 1 Load and Clean Text Data
# Reading dataset (text and emotion columns)
df = pd.read_csv('train.txt', sep=';', header=None, names=['text', 'emotion'])

# Basic text cleaning: removing extra spaces and converting to lowercase
df['clean_text'] = df['text'].str.strip().str.lower()

X = df['clean_text']
y = df['emotion']

# 2 Perform Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print(f"Data Split Done! Training samples: {len(X_train)}, Testing samples: {len(X_test)}")

#  3 Apply both CountVectorizer and TfidfVectorizer
#  CountVectorizer
cv = CountVectorizer()
X_train_cv = cv.fit_transform(X_train)
X_test_cv = cv.transform(X_test)

# TfidfVectorizer
tfidf = TfidfVectorizer()
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

#  4 Train MultinomialNB on both.
# Model for CountVectorizer
model_cv = MultinomialNB()
model_cv.fit(X_train_cv, y_train)

# Model for TfidfVectorizer
model_tfidf = MultinomialNB()
model_tfidf.fit(X_train_tfidf, y_train)

# 5 Compare Accuracy
y_pred_cv = model_cv.predict(X_test_cv)
acc_cv = accuracy_score(y_test, y_pred_cv)

y_pred_tfidf = model_tfidf.predict(X_test_tfidf)
acc_tfidf = accuracy_score(y_test, y_pred_tfidf)

print("\n=== Model Accuracy Comparison ===")
print(f"CountVectorizer Accuracy : {acc_cv * 100:.2f}%")
print(f"TfidfVectorizer Accuracy  : {acc_tfidf * 100:.2f}%")

#  6 Save the Best Vectorizer and the Best Model using joblib.
if acc_tfidf >= acc_cv:
    print("\nResult: TF-IDF performed better. Saving TF-IDF model & vectorizer...")
    best_vectorizer = tfidf
    best_model = model_tfidf
else:
    print("\nResult: CountVectorizer performed better. Saving BoW model & vectorizer...")
    best_vectorizer = cv
    best_model = model_cv

# Save models using joblib
joblib.dump(best_vectorizer, 'best_vectorizer.pkl')
joblib.dump(best_model, 'best_emotion_model.pkl')

print("Files saved successfully as 'best_vectorizer.pkl' and 'best_emotion_model.pkl'!")
```
